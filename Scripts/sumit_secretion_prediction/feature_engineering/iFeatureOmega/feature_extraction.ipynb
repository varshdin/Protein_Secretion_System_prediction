```python
train_path = 'C:/Users/sumit/OneDrive/Documenten/Thesis/comp-ss/Results/sumit_secretion_prediction/phase0_dataset_prep/Train_Test/train.fasta'

standard_aa = set("ACDEFGHIKLMNPQRSTVWY")
bad_chars_found = set()
bad_char_sequences = []

with open(train_path) as f:
    current_id = None
    for line in f:
        line = line.strip()
        if line.startswith('>'):
            current_id = line
        elif line:
            extra = set(line) - standard_aa
            if extra:
                bad_chars_found |= extra
                bad_char_sequences.append((current_id, extra))

print("Non-standard characters found across whole file:", bad_chars_found)
print("Number of sequences affected:", len(bad_char_sequences))
if bad_char_sequences:
    print("First 5 affected IDs:", bad_char_sequences[:5])
```

```python
short_seqs = []

with open(train_path) as f:
    current_id = None
    for line in f:
        line = line.strip()
        if line.startswith('>'):
            current_id = line
        elif line:
            if len(line) < 100:
                short_seqs.append((current_id, len(line)))

print("Sequences shorter than 100 residues:", len(short_seqs))
if short_seqs:
    print("First 5 short ones (ID, length):", short_seqs[:5])
    lengths_only = [x[1] for x in short_seqs]
    print("Shortest length found:", min(lengths_only))
```

```python
from collections import defaultdict

class_short_counts = defaultdict(int)
class_total_counts = defaultdict(int)

with open(train_path) as f:
    current_id = None
    for line in f:
        line = line.strip()
        if line.startswith('>'):
            current_id = line
            class_name = current_id[1:].split('-')[0]
            class_total_counts[class_name] += 1
        elif line:
            if len(line) < 100:
                class_short_counts[class_name] += 1

for cls in class_total_counts:
    total = class_total_counts[cls]
    short = class_short_counts.get(cls, 0)
    print(f"{cls}: {short}/{total} short ({100*short/total:.1f}%)")
```
